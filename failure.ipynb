```
from flask import Flask, jsonify, request

app = Flask(__name__)
port = 3000

# رنگ‌ها
COLORS = {
    "normal": "#2ECC71",   # سبز
    "warning": "#FFC300",  # زرد
    "failure": "#C70039"   # قرمز
}

# لیست خطاها با تقسیم‌بندی
failures = [
    {"Label": "Flow Instability", "Color": COLORS["warning"], "Type": "AI"},
    {"Label": "Pressure Anomaly", "Color": COLORS["failure"], "Type": "AI"},
    {"Label": "Chemical Composition", "Color": COLORS["normal"], "Type": "AI"},
    {"Label": "Temperature Anomaly", "Color": COLORS["normal"], "Type": "AI"},
    {"Label": "Fluid Level Anomaly", "Color": COLORS["normal"], "Type": "AI"},
    {"Label": "Viscosity Anomaly", "Color": COLORS["normal"], "Type": "AI"},
    {"Label": "Gas Analysis", "Color": COLORS["normal"], "Type": "AI"},
    {"Label": "Flow Rate", "Color": COLORS["normal"], "Type": "AI"}
]

# API با فیلتر قابل پارامتر شدن
@app.route("/api/failures", methods=["GET"])
def get_failures():
    color_filter = request.args.get("color")   # مثلاً warning، failure، normal
    type_filter = request.args.get("type")     # مثلاً AI

    filtered = failures
    if color_filter:
        filtered = [f for f in filtered if f["Color"] == COLORS.get(color_filter, "")]

    if type_filter:
        filtered = [f for f in filtered if f["Type"].lower() == type_filter.lower()]

    return jsonify({"Items": filtered})

if __name__ == "__main__":   # ✅ اینجا اصلاح شد
    print(f"API running at http://localhost:{port}")
    app.run(port=port)
```
